# **Import & Load Data**

```python
import pandas as pd

pd.set_option('display.width', 200)
pd.set_option('display.max_columns', None)

df = pd.read_csv('/content/paket_a_transaksi_pos_2024_2025.csv')
df.info()
```

# **Task 1**

```python
print("Tipe data sebelum:", df['harga_satuan'].dtype)

df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(',', '.', regex=False)
df['harga_satuan'] = df['harga_satuan'].astype(float)

print("Tipe data sesudah:", df['harga_satuan'].dtype)
```

```python
print("Jumlah negatif sebelum:", (df['jumlah'] < 0).sum())

df.loc[df['jumlah'] < 0, 'jumlah'] = df.loc[df['jumlah'] < 0, 'jumlah'].abs()

print("Jumlah negatif sesudah:", (df['jumlah'] < 0).sum())
```

```python
print("NaN kode_kasir sebelum:", df['kode_kasir'].isna().sum())

df.loc[df['kode_kasir'].isna(), 'kode_kasir'] = 'K_UNKNOWN'

print("NaN kode_kasir sesudah:", df['kode_kasir'].isna().sum())
```

```python
# kolom: id_transaksi (0), tanggal (1), kode_gerai (2), jumlah (5)
df.iloc[:5, [0, 1, 2, 5]]
```

# **Task 2**

```python
df['Total_Kotor'] = df['jumlah'] * df['harga_satuan']
df['Nominal_Diskon'] = df['Total_Kotor'] * (df['diskon_persen'] / 100)
```

```python
df['Nominal_Diskon'] = df['Total_Kotor'] * (df['diskon_persen'] / 100)
```

```python
df['Total_Bersih'] = (df['Total_Kotor'] - df['Nominal_Diskon']) * 1.11

df[['id_transaksi', 'harga_satuan', 'jumlah', 'diskon_persen',
    'Total_Kotor', 'Nominal_Diskon', 'Total_Bersih']].head()
```

# **Task 5**

```python
df_selesai = df[df['status'] == 'SELESAI']

laporan_gerai = (df_selesai.groupby('kode_gerai')
                 .agg(Jumlah_Transaksi=('id_transaksi', 'count'),
                      Total_Unit_Terjual=('jumlah', 'sum'),
                      Total_Omzet_Bersih=('Total_Bersih', 'sum'),
                      Rata_Rata_Transaksi=('Total_Bersih', 'mean'))
                 .sort_values('Total_Omzet_Bersih', ascending=False)
                 .round(2))
```

```python
total_gerai = df.groupby('kode_gerai')['Total_Bersih'].transform('sum')
df['Kontribusi_Gerai_Persen'] = (df['Total_Bersih'] / total_gerai) * 100

df[['id_transaksi', 'kode_gerai', 'Total_Bersih', 'Kontribusi_Gerai_Persen']].head(10)
```
