```python
import csv
import pandas as pd
```

```python
nama_file = "paket_a_transaksi_pos_2024_2025.csv.xlsx"
nama_sheet = "paket_a_transaksi_pos_2024_2025"
status_valid = ["SELESAI"]
batas_jumlah_wajar = 100
```

```python
mentah = pd.read_excel(nama_file, sheet_name=nama_sheet, header=None, dtype=str)
baris_teks = mentah[0].tolist()
nama_kolom = baris_teks[0].split(",")
isi_data = list(csv.reader(baris_teks[1:]))
data = pd.DataFrame(isi_data, columns=nama_kolom)
data = data.replace(r"^\s*$", pd.NA, regex=True)
```

```python
bersih = data.drop_duplicates().reset_index(drop=True)
bersih["kode_kasir"] = bersih["kode_kasir"].fillna("TIDAK DIKETAHUI")
bersih["status"] = bersih["status"].str.strip().str.upper()
```

```python
format_iso = bersih["tanggal"].str.match(r"^\d{4}-\d{2}-\d{2}$")
tanggal_iso = pd.to_datetime(bersih.loc[format_iso, "tanggal"], format="%Y-%m-%d")
tanggal_lainnya = pd.to_datetime(
    bersih.loc[~format_iso, "tanggal"].str.replace("/", "-", regex=False),
    format="%d-%m-%Y",
)
bersih["tanggal"] = pd.concat([tanggal_iso, tanggal_lainnya]).sort_index()
bersih["tahun"] = bersih["tanggal"].dt.year
```

```python
import re

def ubah_harga(nilai):
    teks = str(nilai).strip()
    if "," in teks:
        teks = teks.replace(".", "").replace(",", ".")
    elif re.fullmatch(r"\d{1,3}(\.\d{3})+", teks):
        teks = teks.replace(".", "")
    return float(teks)

bersih["harga_satuan"] = bersih["harga_satuan"].apply(ubah_harga)
bersih["jumlah"] = pd.to_numeric(bersih["jumlah"]).abs()
bersih["diskon_persen"] = pd.to_numeric(bersih["diskon_persen"])
bersih["jumlah_tidak_wajar"] = bersih["jumlah"] > batas_jumlah_wajar
```

# **Hitung pendapatan per tahun**

```python
bersih["pendapatan_bersih"] = (
    bersih["jumlah"] * bersih["harga_satuan"] * (1 - bersih["diskon_persen"] / 100)
)
transaksi_valid = bersih[
    bersih["status"].isin(status_valid) & ~bersih["jumlah_tidak_wajar"]
]
```

```python
pendapatan_per_tahun = (
    transaksi_valid.groupby("tahun")["pendapatan_bersih"]
    .sum()
    .loc[[2024, 2025]]
)
for tahun, total in pendapatan_per_tahun.items():
    print(f"Pendapatan tahun {tahun}: Rp {total:,.0f}")
```
